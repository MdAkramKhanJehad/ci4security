# CodeQL: RQ3 library-type effect estimation

This restart-executable notebook preserves the four library-type contrasts from the legacy analysis, retains all eligible source rows, and applies PSM, standardized logistic regression, doubly robust AIPW, overlap and balance diagnostics, row and APK-cluster bootstraps, and five refutation or sensitivity checks.

```python
from pathlib import Path
import sys
from IPython.display import display

REPO_ROOT = Path.cwd().resolve()
while REPO_ROOT != REPO_ROOT.parent and not (REPO_ROOT / 'causal_analysis').is_dir():
    REPO_ROOT = REPO_ROOT.parent
if not (REPO_ROOT / 'causal_analysis').is_dir():
    raise RuntimeError('Run this notebook from the ci4security repository')
if str(REPO_ROOT) not in sys.path:
    sys.path.insert(0, str(REPO_ROOT))

from causal_analysis.shared.rq3_library_types import run_rq3_library_type_analysis
```

```python
TOOL_NAME = 'CodeQL'
DATA_PATH = REPO_ROOT / 'causal_analysis/data/codeql/alerts_with_lib_category_and_apk_size_codeql.csv'
OUTPUT_DIR = REPO_ROOT / 'causal_analysis/rq3_effect_estimation/results/codeql/library_types'
BOOTSTRAP_SIMULATIONS = 1000
REFUTER_SIMULATIONS = 200
RANDOM_SEED = 20260915
```

```python
analysis = run_rq3_library_type_analysis(
    tool_name=TOOL_NAME,
    source_path=DATA_PATH,
    output_dir=OUTPUT_DIR,
    bootstrap_simulations=BOOTSTRAP_SIMULATIONS,
    refuter_simulations=REFUTER_SIMULATIONS,
    seed=RANDOM_SEED,
)
print(f'Results written to {OUTPUT_DIR}')
```

```python
display(analysis['contrast_summary'])
display(analysis['estimator_results'])
display(analysis['overlap'])
display(analysis['balance'])
display(analysis['matching'])
display(analysis['refuters'])
display(analysis['software_validation'])
```
